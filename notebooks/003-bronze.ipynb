Ler os CSVs do Lading

```python
caminho_landing = "/Volumes/workspace/landing/dados"

df_categoria = spark.read.option("inferSchema", "true").option("header", "true").csv(f"{caminho_landing}/categoria")
df_marca     = spark.read.option("inferSchema", "true").option("header", "true").csv(f"{caminho_landing}/marca")
df_modelo    = spark.read.option("inferSchema", "true").option("header", "true").csv(f"{caminho_landing}/modelo")
df_estado    = spark.read.option("inferSchema", "true").option("header", "true").csv(f"{caminho_landing}/estado")
df_cidade    = spark.read.option("inferSchema", "true").option("header", "true").csv(f"{caminho_landing}/cidade")
df_agencia   = spark.read.option("inferSchema", "true").option("header", "true").csv(f"{caminho_landing}/agencia")
df_cliente   = spark.read.option("inferSchema", "true").option("header", "true").csv(f"{caminho_landing}/cliente")
df_carro     = spark.read.option("inferSchema", "true").option("header", "true").csv(f"{caminho_landing}/carro")
df_reserva   = spark.read.option("inferSchema", "true").option("header", "true").csv(f"{caminho_landing}/reserva")
df_pagamento = spark.read.option("inferSchema", "true").option("header", "true").csv(f"{caminho_landing}/pagamento")
```

Adicionar metadados

```python
from pyspark.sql.functions import current_timestamp, lit

df_categoria = df_categoria.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("categoria.csv"))
df_marca     = df_marca.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("marca.csv"))
df_modelo    = df_modelo.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("modelo.csv"))
df_estado    = df_estado.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("estado.csv"))
df_cidade    = df_cidade.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("cidade.csv"))
df_agencia   = df_agencia.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("agencia.csv"))
df_cliente   = df_cliente.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("cliente.csv"))
df_carro     = df_carro.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("carro.csv"))
df_reserva   = df_reserva.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("reserva.csv"))
df_pagamento = df_pagamento.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("pagamento.csv"))
```

Salvar em Delta Lake no Bronze

```python
df_categoria.write.format('delta').mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.categoria")
df_marca.write.format('delta').mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.marca")
df_modelo.write.format('delta').mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.modelo")
df_estado.write.format('delta').mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.estado")
df_cidade.write.format('delta').mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.cidade")
df_agencia.write.format('delta').mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.agencia")
df_cliente.write.format('delta').mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.cliente")
df_carro.write.format('delta').mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.carro")
df_reserva.write.format('delta').mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.reserva")
df_pagamento.write.format('delta').mode("overwrite").option("overwriteSchema", "true").saveAsTable("bronze.pagamento")
```
