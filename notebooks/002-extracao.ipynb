```python
# Configurações de conexão com o Supabase (PostgreSQL)
jdbc_url = "jdbc:postgresql://aws-1-us-west-2.pooler.supabase.com:5432/postgres?user=postgres.jefhanemiqhetswshoai&password=[YOUR-PASSWORD]"

caminho_landing = "/Volumes/workspace/landing/dados"

# Lista de tabelas para extrair
tabelas = [
    "categoria",
    "marca",
    "modelo",
    "estado",
    "cidade",
    "agencia",
    "cliente",
    "carro",
    "reserva",
    "pagamento"
]

# Extrai cada tabela e salva como CSV no Volume landing
for tabela in tabelas:
    df = spark.read \
        .format("jdbc") \
        .option("url", jdbc_url) \
        .option("dbtable", tabela) \
        .option("driver", "org.postgresql.Driver") \
        .load()
    
    df.write \
        .mode("overwrite") \
        .option("header", "true") \
        .csv(f"{caminho_landing}/{tabela}")
    
    print(f"✅ Tabela '{tabela}' extraída com sucesso!")

print("\n Extração concluída! Todas as tabelas estão no Landing.")
```
